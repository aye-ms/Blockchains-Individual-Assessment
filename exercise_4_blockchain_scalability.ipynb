# Exercise 4: Blockchain Scalability, the Trilemma and Layer-2 Solutions

```python
average_transaction_bytes = 580
average_block_time_seconds = 576
bytes_per_megabyte = 1_000_000
block_sizes_mb = [1, 2, 4]

bitcoin_results = []

for block_size_mb in block_sizes_mb:
    block_size_bytes = block_size_mb * bytes_per_megabyte
    transactions_per_block = (
        block_size_bytes / average_transaction_bytes
    )
    theoretical_tps = (
        transactions_per_block / average_block_time_seconds
    )

    bitcoin_results.append({
        "block_size_mb": block_size_mb,
        "transactions_per_block": transactions_per_block,
        "theoretical_tps": theoretical_tps
    })

print("BITCOIN THEORETICAL THROUGHPUT")
print(
    f"{'Block size':<13} {'Transactions/block':<21} "
    f"{'Block time':<13} {'TPS':<8}"
)

for result in bitcoin_results:
    print(
        f"{result['block_size_mb']} MB{'':<8} "
        f"{result['transactions_per_block']:<21,.2f} "
        f"{average_block_time_seconds} seconds{'':<2} "
        f"{result['theoretical_tps']:<8.2f}"
    )
```
