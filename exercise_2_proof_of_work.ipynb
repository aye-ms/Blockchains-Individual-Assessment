# Exercise 2: Proof of Work - Mining, Difficulty and Probability

```python
import hashlib
import pickle
import time

def mine_block(block_data, difficulty):
    prefix = "0" * difficulty
    nonce = 1
    attempts = 0

    pickled_block = pickle.dumps(block_data)

    start_time = time.time()

    while True:
        block_hash = hashlib.sha256(
            str(nonce).encode() + pickled_block
        ).hexdigest()
        attempts = attempts + 1

        if block_hash.startswith(prefix):
            mining_time = time.time() - start_time
            return nonce, block_hash, attempts, mining_time

        nonce = nonce + 1


def calculate_results(times, attempts):
    average_time = sum(times) / len(times)
    minimum_time = min(times)
    maximum_time = max(times)

    variance = sum(
        (mining_time - average_time) ** 2 for mining_time in times
    ) / (len(times) - 1)
    standard_deviation = variance ** 0.5

    total_attempts = sum(attempts)
    total_time = sum(times)
    throughput = total_attempts / total_time
    average_attempts = total_attempts / len(attempts)

    return (
        average_time, minimum_time, maximum_time, standard_deviation,
        throughput, average_attempts
    )


difficulties = [2, 3, 4, 5]
all_results = []

for difficulty in difficulties:
    mining_times = []
    hash_attempts = []

    print()
    print("DIFFICULTY:", difficulty)

    for block_number in range(1, 31):
        block_data = {
            "block_number": block_number,
            "difficulty": difficulty
        }
        nonce, valid_hash, attempts, mining_time = mine_block(
            block_data, difficulty
        )

        mining_times.append(mining_time)
        hash_attempts.append(attempts)

        print(
            "Block:", block_number,
            "| Time:", round(mining_time, 6), "seconds",
            "| Attempts:", attempts,
            "| Valid hash:", valid_hash
        )

    results = calculate_results(mining_times, hash_attempts)
    theoretical_attempts = 16 ** difficulty

    all_results.append({
        "difficulty": difficulty,
        "average_time": results[0],
        "minimum_time": results[1],
        "maximum_time": results[2],
        "standard_deviation": results[3],
        "throughput": results[4],
        "average_attempts": results[5],
        "theoretical_attempts": theoretical_attempts
    })


print()
print("SUMMARY TABLE")
print(
    f"{'Difficulty':<11} {'Average(s)':<12} {'Minimum(s)':<12} "
    f"{'Maximum(s)':<12} {'Std Dev(s)':<12} {'Hashes/sec':<13} "
    f"{'Avg Attempts':<14} {'Theoretical':<12}"
)

for result in all_results:
    print(
        f"{result['difficulty']:<11} "
        f"{result['average_time']:<12.6f} "
        f"{result['minimum_time']:<12.6f} "
        f"{result['maximum_time']:<12.6f} "
        f"{result['standard_deviation']:<12.6f} "
        f"{result['throughput']:<13.2f} "
        f"{result['average_attempts']:<14.2f} "
        f"{result['theoretical_attempts']:<12}"
    )
```
